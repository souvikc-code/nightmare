# bs

for all student mails in report_bs,
 - if credit.TotalCredits == sum(credit['BD', 'BP', 'DD', 'DP', 'FL', 'HM', 'SE', 'BP_Tc', 'BD_Tc', 'HM_Tc', 'SE_Tc'])
 - report_bs.All_Bsc_Completed == credit.All_Bsc_Completed
 - report_bs.All_Bs_Completed == credit.All_Bs_Completed
 - report_bs.DDCompleted == credit.DDCompleted
 - report_bs.DPCompleted == credit.DPCompleted
 - if credit.TotalCredits >= 142
 - if credit.FL == 32
 - if credit.DD == 27
 - if credit.DP == 27
 - if credit.BD >= 16
 - if credit.BP >= 16
 - if credit.HM >= 8

```python
import pandas as pd
import numpy as np

def check_bs(report_bs: pd.DataFrame, credit: pd.DataFrame) -> pd.DataFrame:

    df = report_bs.merge(credit, on="student_email", how="left", suffixes=("_report", "_credit"))

    results = []

    credit_sum_cols = [
        'BD', 'BP', 'DD', 'DP', 'FL', 'HM', 'SE',
        'BP_Tc', 'BD_Tc', 'HM_Tc', 'SE_Tc'
    ]

    for _, row in df.iterrows():
        failed_checks = []

        # check 1: TotalCredits sum match
        credit_sum = row[credit_sum_cols].sum(skipna=False)
        if row['TotalCredits'] != credit_sum:
            failed_checks.append("check_totalcredits_sum")

        # check 2: All_Bsc_Completed
        if row['All_Bsc_Completed_report'] != row['All_Bsc_Completed_credit']:
            failed_checks.append("check_all_bsc_completed")

        # check 3: All_Bs_Completed
        if row['All_Bs_Completed_report'] != row['All_Bs_Completed_credit']:
            failed_checks.append("check_all_bs_completed")

        # check 4: DDCompleted
        if row['DDCompleted_report'] != row['DDCompleted_credit']:
            failed_checks.append("check_dd_completed")

        # check 5: DPCompleted
        if row['DPCompleted_report'] != row['DPCompleted_credit']:
            failed_checks.append("check_dp_completed")

        # check 6: TotalCredits >= 142
        if row['TotalCredits'] < 142:
            failed_checks.append("check_totalcredits_142")

        # check 7: FL == 32
        if row['FL'] != 32:
            failed_checks.append("check_fl_32")

        # check 8: DD == 27
        #if row['DD'] != 27:
        if not (
          (row['DD'] == 27) or
          (row['DD'] == 25 and row['DDCompleted_credit'] is True) or
          (row['DD'] == 23 and row['DDCompleted_credit'] is True)):
            failed_checks.append("check_dd_27")

        # check 9: DP == 27
        #if row['DP'] != 27:
            #failed_checks.append("check_dp_27")
        if not (
          (row['DP'] == 27) or
          (row['DP'] == 25 and row['DPCompleted_credit'] is True) or
          (row['DP'] == 23 and row['DPCompleted_credit'] is True)):
            failed_checks.append("check_dp_27")

        # check 10: BD >= 16
        if (row['BD'] + row['BD_Tc']) < 16:
            failed_checks.append("check_bd_16")

        # check 11: BP >= 16
        if (row['BP'] + row['BP_Tc']) < 16:
            failed_checks.append("check_bp_16")

        # check 12: HM >=8
        if (row['HM']+ row['HM_Tc']) < 8:
            failed_checks.append("check_hm_8")

        # check 13:

        # final status
        status = "yes" if not failed_checks else failed_checks

        results.append({
            "student_email": row["student_email"],
            "status": status
        })

    return pd.DataFrame(results)
```

```python
result= check_bs(report_bs, credit)
```
